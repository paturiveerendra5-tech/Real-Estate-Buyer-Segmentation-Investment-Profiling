### PART 14: BUYER AGE DISTRIBUTION ANALYSIS

```python
# Step 1: Convert Date of Birth into datetime
clients_eda = clients.copy()

clients_eda["date_of_birth"] = pd.to_datetime(
    clients_eda["date_of_birth"],
    errors="coerce",
    dayfirst=True
)

# Step 2: Calculate buyer age
reference_date = pd.Timestamp.today().normalize()

clients_eda["age"] = (
    reference_date.year
    - clients_eda["date_of_birth"].dt.year
    - (
        (reference_date.month < clients_eda["date_of_birth"].dt.month)
        | (
            (reference_date.month == clients_eda["date_of_birth"].dt.month)
            & (reference_date.day < clients_eda["date_of_birth"].dt.day)
        )
    ).astype(int)
)

# Step 3: Display age statistics
print("BUYER AGE DISTRIBUTION")
print("----------------------")
print(clients_eda["age"].describe())

# Step 4: Create Histogram
plt.figure(figsize=(10, 5))

plt.hist(
    clients_eda["age"].dropna(),
    bins=15,
    edgecolor="black"
)

plt.title("Buyer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Buyers")
plt.tight_layout()

# Step 5: Display chart
plt.show()
```

### PART 15 – CORRELATION ANALYSIS

```python
# Select numerical columns
numeric_data = properties_eda[
    ["floor_area_sqft", "sale_price"]
].copy()

# Calculate correlation
correlation_matrix = numeric_data.corr()

# Display correlation
print("CORRELATION ANALYSIS")
print("--------------------")
print(correlation_matrix)

# Create correlation heatmap
plt.figure(figsize=(8, 5))

plt.imshow(correlation_matrix, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar(label="Correlation")

plt.xticks(range(len(correlation_matrix.columns)),
           correlation_matrix.columns, rotation=45)

plt.yticks(range(len(correlation_matrix.columns)),
           correlation_matrix.columns)

plt.title("Property Features Correlation")

# Display correlation values
for i in range(len(correlation_matrix.columns)):
    for j in range(len(correlation_matrix.columns)):
        plt.text(
            j, i,
            f"{correlation_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.tight_layout()
plt.show()
```
